```python
def load_dataset(size=None, random_state=None, force_balanced=True):
    """Loads dataset"""
    mat_file = "BigDigits.mat"
    mat = matlab.loadmat(mat_file, squeeze_me=True)
    data = mat["data"]
    labels = mat["labs"] - 1

    # Convert labels to 0 and 1 for binary classification
    new_labels = [0 if label == 0 else 1 for label in labels]
    labels = np.array(new_labels)

    # Force balance of the dataset
    if force_balanced:
        class_0 = data[labels == 0]
        class_1 = data[labels == 1]
        min_class_size = min(len(class_0), len(class_1))

        class_0_labels = np.zeros(min_class_size, dtype=int)
        class_1_labels = np.ones(min_class_size, dtype=int)

        class_0 = resample(class_0, replace=False, n_samples=min_class_size, random_state=random_state)
        class_1 = resample(class_1, replace=False, n_samples=min_class_size, random_state=random_state)

        data = np.vstack((class_0, class_1))
        labels = np.hstack((class_0_labels, class_1_labels))

    # Select a subset of data if size is provided
    if size:
        half_size = size // 2
        data = np.vstack((data[:half_size], data[-half_size:]))
        labels = np.hstack((labels[:half_size], labels[-half_size:]))

    # Randomize the data and labels
    np.random.seed(random_state)
    indices = np.arange(data.shape[0])
    np.random.shuffle(indices)
    data = data[indices]
    labels = labels[indices]

    return data, labels
```
